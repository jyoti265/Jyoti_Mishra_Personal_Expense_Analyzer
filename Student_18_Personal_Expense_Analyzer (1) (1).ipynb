## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# TODO: Create calculated column(s)
# TODO: Filter records using a condition
# TODO: Sort records and display the top/bottom result

#create two calculated column

df["expense_level"] = np.where(df["Amount"] > 200, "High", "Normal")


df["remaining_budget"] = 10000 - df["Amount"]
print(df.head())  



df = df.drop(columns=["\nremaining_budget"], errors="ignore")
print(df.columns.tolist())
display(df.head())


#filter records
high_expenses = df[df["Amount"]> 200]

display(high_expenses)

#sort amount from highest to lowest

#top result
sorted_expenses = df.sort_values(by="Amount", ascending=False)
display(sorted_expenses.head(5))
#bottom result
display(sorted_expenses.tail(5))

#sort amount from lowest to highest
#top result
lowest_expenses = df.sort_values(by="Amount", ascending=True)
display(lowest_expenses.head(5))
#bottom result
display(sorted_expenses.tail(5))

```

## 6. Your individual project challenge
**Challenge:** Calculate total expenses; category-wise totals and averages; identify largest expense category.

Write your solution below and explain what the result means.


```python
# TODO: Solve your assigned challenge



total_expenses = df["Amount"].sum()

print("Total Expenses:", total_expenses)



category_total = df.groupby("Category")["Amount"].sum()

print("\nCategory-wise Total:")
print(category_total)



category_average = df.groupby("Category")["Amount"].mean()

print("\nCategory-wise Average:")
print(category_average)



largest_category = category_total.idxmax()

print("\nLargest Expense Category:", largest_category)
```
